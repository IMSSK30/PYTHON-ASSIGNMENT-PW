# Regional Sales Performance

4.Calculate the total sales generated by each region.
Then identify:
-Which region generated the highest revenue?

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(region_sales)

print('Highest Revenue Region:', region_sales.idxmax())
```

# Product Performance Analysis

5.Calculate the total quantity sold for each product.
Which product appears to be the most popular?

```python
product_sales = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
print(product_sales)

print('Most Popular Product:', product_sales.idxmax())
```

# Pivot Table Creation

```python
Q7.Create a pivot table showing:
Rows → Region
Columns → Product
Values → Total Sales
Which region sells the most laptops?
```

```python
pivot_table = pd.pivot_table(df,
                             values='Sales',
                             index='Region',
                             columns='Product',
                             aggfunc='sum')

print(pivot_table)
```

```python
print("Region with highest Laptop sales:", pivot_table['Laptop'].idxmax())
```

# Profitability Analysis

Q8.Calculate the average profit for each product category.
Which category is most profitable?

```python
profit_analysis = df.groupby('Category')['Profit'].mean().sort_values(ascending=False)
print(profit_analysis)

print('Most Profitable Category:', profit_analysis.idxmax())
```

# Discount Impact Study

Q9.Analyze the relationship between discount and profit.

# End-to-End Data Wrangling Task

Q10.Perform the following workflow:
-Filter the dataset where Sales > 3000
-Group data by Region and Product
-Calculate total sales
-Reshape the result into a pivot table
-Explain the business insights obtained from this analysis.

```python
# Step 1: Filter
filtered_df = df[df['Sales'] > 3000]

# Step 2 & 3: Group and aggregate
grouped_df = filtered_df.groupby(['Region', 'Product'])['Sales'].sum()

# Step 4: Pivot table
pivot_result = grouped_df.unstack()

print(pivot_result)
```

Insight:
This shows high-value sales across regions and products, helping identify key revenue drivers.

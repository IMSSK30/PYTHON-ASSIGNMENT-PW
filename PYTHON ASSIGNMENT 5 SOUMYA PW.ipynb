## Q1. Handle Missing Values (Basic Cleaning)

```python
import pandas as pd
import numpy as np

df = pd.read_csv(r'C:\Users\soumy\Downloads\employee_productivity_dataset.csv')

print("Missing Values Before:")
print(df.isnull().sum())

df['Age'].fillna(df['Age'].median(), inplace=True)
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
df['Hours_Worked_Per_Week'].fillna(df['Hours_Worked_Per_Week'].median(), inplace=True)
df['Performance_Score'].fillna(df['Performance_Score'].mean(), inplace=True)

print("Dataset After Handling Missing Values:")
print(df.head())
```

Median is used for Age and Hours_Worked_Per_Week because it handles outliers well.
Mean is used for Salary and Performance_Score because they are continuous numerical values.

## Q2. Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Gender'] = le.fit_transform(df['Gender'])
df['Department'] = le.fit_transform(df['Department'])

df[['Gender', 'Department']].head()
```

Label Encoding converts categorical text values into numerical form.

## Q3. One-Hot Encoding

```python
df = pd.get_dummies(df, columns=['Work_Mode', 'Location'])

print(df.head())

print("Total Columns After Encoding:", len(df.columns))
```

One-Hot Encoding creates separate binary columns for each category.

## Q4. Normalization (Min-Max Scaling)

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()

df[['Salary', 'Hours_Worked_Per_Week']] = minmax.fit_transform(df[['Salary', 'Hours_Worked_Per_Week']])

print(df[['Salary', 'Hours_Worked_Per_Week']].head())
```

Min-Max Scaling converts values between 0 and 1.

## Q5. Standardization (Scaling)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Age', 'Projects_Completed']] = scaler.fit_transform(df[['Age', 'Projects_Completed']])

print(df[['Age', 'Projects_Completed']].head())
```

Standardization transforms data so that:
Mean = 0
Standard Deviation = 1

## Q6. Compare Scaling Methods

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

salary_data = df[['Salary']]

minmax = MinMaxScaler()
salary_minmax = minmax.fit_transform(salary_data)

standard = StandardScaler()
salary_standard = standard.fit_transform(salary_data)

comparison = pd.DataFrame({
    'Original_Salary': df['Salary'],
    'MinMax_Scaled': salary_minmax.flatten(),
    'Standard_Scaled': salary_standard.flatten()})

print(comparison.head())
```

MinMaxScaler scales values between 0 and 1.
StandardScaler centers values around mean 0.

## Q7. Build Preprocessing Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

numerical_cols = ['Age', 'Salary', 'Hours_Worked_Per_Week',
                  'Projects_Completed', 'Performance_Score']

pipeline = Pipeline([('scaler', StandardScaler())])

transformed_data = pipeline.fit_transform(df[numerical_cols])

print("Pipeline Created Successfully")
```

The pipeline should now only scale numerical columns because we have already done encode to categorical columns in previous question.
--ColumnTransformer applies different preprocessing steps to different columns.
--Pipeline automates preprocessing workflow.

## Q8. Apply Pipeline

```python
transformed_data = pipeline.fit_transform(df[numerical_cols])

print(transformed_data)

print("Shape of Final Dataset:")
print(transformed_data.shape)
```

The pipeline transforms numerical and categorical data together into ML-ready format.

## Q9. Why is scaling important in Python?

Scaling is important because machine learning algorithms perform better when features are on a similar scale.
It:
--Improves model accuracy
--Speeds up training
--Prevents large-value features from dominating smaller-value features
--Helps distance-based algorithms work properly

## 10. Why do we convert categorical data into numerical form?

Machine learning models work with mathematical calculations and cannot directly understand text values. Converting categorical data into numerical form:
--Makes data understandable for ML models
--Improves processing efficiency
--Helps algorithms perform accurate predictions

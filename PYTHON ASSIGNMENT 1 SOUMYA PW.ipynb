**Question 5: Write a Python program to check whether a number is even or odd.**

```python
num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

**Question 6: Write a program to print numbers from 1 to 10 using a loop.**

```python
i = 1
while i <= 10:
    print(i)
    i = i + 1
```

**Question 7: Create a list of five numbers and print the maximum number.**

```python
numbers = [10, 20, 5, 40, 15]
max_num = numbers[0]
for n in numbers:
    if n > max_num:
        max_num = n
print(max_num)
```

**Question 8: Write a program to remove duplicate values from a list using a set.**

```python
lst = [1, 2, 2, 3, 4, 4, 5]
new_list = list(set(lst))
print(new_list)
```

**Question 9: Write a function that returns the square of a number.**

**Question 10: Write a Python program to count how many times a number appears in a list.**

```python
lst = [2, 3, 4, 2, 5, 2]
num = int(input("Enter number: "))
count = 0
for i in lst:
    if i == num:
        count = count + 1
print(count)
```

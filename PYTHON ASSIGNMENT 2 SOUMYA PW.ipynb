**Q1. Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows along with the dataset information.**

```python
import pandas as pd

df = pd.read_csv(r'C:\Users\soumy\Downloads\retail_dataset .csv')

print(df.head())
print(df.info())
```

**Q2. Inspect the dataset and identify missing values in each column. Display the total number of missing values present in the dataset.**

```python
print(df.isnull().sum())

total_missing = df.isnull().sum().sum()
print('Total missing values:', total_missing)
```

**Q3. Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.**

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
df['Price'] = df['Price'].fillna(df['Price'].mean())
```

**Q4. Remove all rows where the Product Category or Region column contains missing values.**

```python
df = df.dropna(subset=['Product Category', 'Region'])
```

**Q5. Create a new column called Revenue using NumPy, where: Revenue = Quantity × Price Display the updated dataset with the new column.**

```python
import numpy as np

df['Revenue'] = df['Quantity'] * df['Price']

print(df.head())
```

**Q6. Using NumPy, calculate the total revenue generated from all transactions in the dataset.**

```python
total_revenue = df['Revenue'].sum()
print("Total Revenue:", total_revenue)
```

**Q7. Group the dataset by Product Category and calculate the total revenue generated by each category.**

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum()

print(category_revenue)
```

**Q8. Based on the revenue calculated for each product category, identify the top 3 categories with the highest revenue
and the bottom 3 categories with the lowest revenue.**

```python
sorted_data = category_revenue.sort_values()

print("Bottom 3 Categories:")
print(sorted_data.head(3))

print("Top 3 Categories:")
print(sorted_data.tail(3))
```

**Q9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the region with the highest revenue
and the region with the lowest revenue.**

```python
region_revenue = df.groupby('Region')['Revenue'].sum()

print(region_revenue)

print("Highest Revenue Region:", region_revenue.idxmax())
print("Lowest Revenue Region:", region_revenue.idxmin())
```

**Q10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the total monthly revenue.
Also, calculate the mean, median, and standard deviation of the Revenue column using NumPy.**

```python
import pandas as pd

# Convert Date column (fix for DD-MM-YYYY format)
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

# Remove rows where Date could not be converted (optional b
df = df.dropna(subset=['Date'])

# Extract month
df['Month'] = df['Date'].dt.month

# Group by month and calculate total revenue
monthly_revenue = df.groupby('Month')['Revenue'].sum()

print("Monthly Revenue:")
print(monthly_revenue)

# Calculate statistics
mean_value = df['Revenue'].mean()
median_value = df['Revenue'].median()
std_value = df['Revenue'].std()

print("Mean:", mean_value)
print("Median:", median_value)
print("Standard Deviation:", std_value)
```
